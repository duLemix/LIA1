```python
# Criando dados para análise estatística
np.random.seed(42)  # Para reproduzibilidade
dados_vendas = np.random.normal(1000, 200, 100)  # 100 vendas com média 1000 e desvio 200

print("Dados de vendas (primeiros 10 valores):")
print(dados_vendas[:10])
print()

# Estatísticas descritivas
print("=== ESTATÍSTICAS DESCRITIVAS ===")
print(f"Média: {np.mean(dados_vendas):.2f}")
print(f"Mediana: {np.median(dados_vendas):.2f}")
print(f"Desvio padrão: {np.std(dados_vendas):.2f}")
print(f"Variância: {np.var(dados_vendas):.2f}")
print(f"Mínimo: {np.min(dados_vendas):.2f}")
print(f"Máximo: {np.max(dados_vendas):.2f}")
print()

# Percentis
print("=== PERCENTIS ===")
print(f"25º percentil: {np.percentile(dados_vendas, 25):.2f}")
print(f"50º percentil (mediana): {np.percentile(dados_vendas, 50):.2f}")
print(f"75º percentil: {np.percentile(dados_vendas, 75):.2f}")
print()

# Soma e produto
print("=== AGREGAÇÕES ===")
print(f"Soma total: {np.sum(dados_vendas):.2f}")
print(f"Produto dos 5 primeiros: {np.prod(dados_vendas[:5]):.2e}")
print(f"Soma acumulada (primeiros 5): {np.cumsum(dados_vendas[:5])}")
```

```python
# Criando um array para manipulação
arr = np.arange(12)
print("Array original:")
print(f"Valores: {arr}")
print(f"Forma: {arr.shape}")
print()

# Reshape - mudando a forma
arr_2d = arr.reshape(3, 4)
print("Array 2D (3x4):")
print(arr_2d)
print(f"Forma: {arr_2d.shape}")
print()

# Reshape para 3D
arr_3d = arr.reshape(2, 2, 3)
print("Array 3D (2x2x3):")
print(arr_3d)
print(f"Forma: {arr_3d.shape}")
print()

# Flatten - achatar o array
arr_flat = arr_3d.flatten()
print("Array achatado:")
print(f"Valores: {arr_flat}")
print(f"Forma: {arr_flat.shape}")
print()
```

```python
# Transpose - transpor matriz
matriz = np.array([[1, 2, 3], [4, 5, 6]])
print("Matriz original:")
print(matriz)
print(f"Forma: {matriz.shape}")
print()

matriz_t = matriz.T
print("Matriz transposta:")
print(matriz_t)
print(f"Forma: {matriz_t.shape}")
print()

# Concatenação
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
concat_h = np.concatenate([arr1, arr2])
print(f"Concatenação horizontal: {concat_h}")

# Stack vertical
stack_v = np.vstack([arr1, arr2])
print("Stack vertical:")
print(stack_v)
```

```python
# Criando um DataFrame com dados "sujos" para demonstrar limpeza
import numpy as np

dados_sujos = {
    'Nome': ['josé','João', 'Ana', 'PEDRO', 'maria', 'Carlos', ''],
    'Idade': [80, 28, 24, 35, None, 45, 30],
    'Salario': [7000, 5000, 6000, None, 4500, 7000, 5500],
    'Cidade': ['Brasília', 'São Paulo', 'Rio de Janeiro', 'Belo Horizonte', 'Salvador', 'Curitiba', 'Fortaleza']
}

df_sujo = pd.DataFrame(dados_sujos)
print("DataFrame com dados 'sujos':")
df_sujo
```

```python
# Verificando dados faltantes
print("Dados faltantes por coluna:")
print(df_sujo.isnull().sum())
print()
```

```python
# Limpando os dados
df_limpo = df_sujo.copy()

# 1. Removendo linhas com nome vazio
df_limpo = df_limpo[df_limpo['Nome'] != '']

# 2. Padronizando nomes (primeira letra maiúscula)
df_limpo['Nome'] = df_limpo['Nome'].str.title()

# 3. Preenchendo valores faltantes
df_limpo['Idade'] = df_limpo['Idade'].fillna(df_limpo['Idade'].median())
df_limpo['Salario'] = df_limpo['Salario'].fillna(df_limpo['Salario'].mean())
print("DataFrame após limpeza:")
df_limpo
```

```python
# Criando nova coluna baseada em outra
df_limpo['Faixa_Etaria'] = df_limpo['Idade'].apply(lambda x: 'Jovem' if x < 30 else 'Adulto' if x < 50 else 'Maduro')
df_limpo['Salario_Categoria'] = pd.cut(df_limpo['Salario'], bins=[0, 5000, 6000, 10000], labels=['Baixo', 'Médio', 'Alto'])

print("DataFrame com novas colunas:")
df_limpo
```

```python
# Criando um DataFrame mais robusto para demonstrar groupby
np.random.seed(42)

dados_vendas = {
    'Vendedor': ['Ana', 'João', 'Pedro', 'Ana', 'João', 'Pedro', 'Ana', 'João', 'Pedro', 'Ana'] * 3,
    'Produto': ['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'] * 6,
    'Regiao': ['Norte', 'Sul', 'Nordeste', 'Centro-Oeste', 'Sudeste'] * 6,
    'Vendas': np.random.randint(1000, 10000, 30),
    'Quantidade': np.random.randint(1, 20, 30)
}

df_vendas_completo = pd.DataFrame(dados_vendas)
print("DataFrame de vendas:")
df_vendas_completo.head(10)
```

```python
# Agrupamento simples - vendas por vendedor
vendas_por_vendedor = df_vendas_completo.groupby('Vendedor')['Vendas'].sum()
print("Vendas totais por vendedor:")
vendas_por_vendedor
```

```python
# Múltiplas agregações
agregacoes = df_vendas_completo.groupby('Vendedor')['Vendas'].agg(['sum', 'mean', 'count'])
print("Múltiplas agregações por vendedor:")
agregacoes
```

```python
# Agrupamento por múltiplas colunas
vendas_por_vendedor_regiao = df_vendas_completo.groupby(['Vendedor', 'Regiao'])['Vendas'].sum()
print("Vendas por vendedor e região:")
vendas_por_vendedor_regiao
```

```python
# Pivot table
pivot_vendas = df_vendas_completo.pivot_table(
    values='Vendas',
    index='Vendedor',
    columns='Produto',
    aggfunc='sum',
    fill_value=0
)
print("Pivot table - Vendas por Vendedor x Produto:")
pivot_vendas
```

```python
# Estatísticas descritivas por grupo
print("Estatísticas descritivas por região:")
df_vendas_completo.groupby('Regiao')['Vendas'].describe()
```

```python
# Salvando o DataFrame em CSV
df_vendas_completo.to_csv('vendas_exemplo.csv', index=False)
print("✅ Arquivo 'vendas_exemplo.csv' salvo com sucesso!")

# Lendo o arquivo CSV
df_lido = pd.read_csv('vendas_exemplo.csv')
print("\n📂 Dados lidos do arquivo CSV:")
df_lido.head()
```

## Desafio 2: Limpeza de Dados

Crie um `DataFrame` com dados "problemáticos" e implemente:
- Remoção de _duplicates_
- Tratamento de valores ausentes
- Padronização de _strings_
- Criação de novas colunas derivadas

```python
n = 30

df_problemas = pd.DataFrame({
    'Vendedor': np.random.choice(['Ana', 'João', 'Pedro'], n),
    'Produto': np.random.choice(['Notebook', 'Mouse', 'Teclado', 'Monitor', 'Headset'], n),
    'Regiao': np.random.choice(['Norte', 'Sul', 'Nordeste', 'Centro-Oeste', 'Sudeste'], n),
    'Vendas': np.random.randint(1000, 10000, n),
    'Quantidade': np.random.randint(1, 20, n)
})

dup_idx = np.random.choice(df_problemas.index, size=3, replace=False)
df_problemas = pd.concat([df_problemas, df_problemas.loc[dup_idx]], ignore_index=True)

nan_idx = np.random.choice(df_problemas.index, size=5, replace=False)
df_problemas.loc[nan_idx[:2], 'Vendedor'] = np.nan
df_problemas.loc[nan_idx[2:4], 'Produto'] = np.nan
df_problemas.loc[nan_idx[4], 'Vendas'] = np.nan

str_idx = np.random.choice(df_problemas.dropna().index, size=4, replace=False)
df_problemas.loc[str_idx[0], 'Vendedor'] = '  ' + df_problemas.loc[str_idx[0], 'Vendedor'] + '  '
df_problemas.loc[str_idx[1], 'Produto'] = df_problemas.loc[str_idx[1], 'Produto'].upper()
df_problemas.loc[str_idx[2], 'Regiao'] = df_problemas.loc[str_idx[2], 'Regiao'].lower()
df_problemas.loc[str_idx[3], 'Regiao'] = '  ' + df_problemas.loc[str_idx[3], 'Regiao'] + ' '

df_problemas
```

```python
duplicates = df_problemas[df_problemas.duplicated(keep='first')]
print(f"Linhas duplicadas encontradas ({len(duplicates)}):")
print(duplicates)
print()

print(f"Linhas antes da remoção de duplicatas: {len(df_problemas)}")
df_problemas = df_problemas.drop_duplicates().reset_index(drop=True)
print(f"Linhas após a remoção de duplicates: {len(df_problemas)}")
```

```python
valores_ausentes = df_problemas.isnull().sum()
print("Valores ausentes por coluna:")
print(valores_ausentes[valores_ausentes > 0])
print()

print(f"Linhas antes do tratamento: {len(df_problemas)}")
df_problemas = df_problemas.dropna().reset_index(drop=True)
print(f"Linhas após o tratamento: {len(df_problemas)}")
```

```python
colunas_str = ['Vendedor', 'Produto', 'Regiao']

print("Valores com espaços extras ou caso inconsistente:")
for col in colunas_str:
    inconsistente = df_problemas[col][
        (df_problemas[col] != df_problemas[col].str.strip()) |
        (df_problemas[col] != df_problemas[col].str.title())
    ]
    if not inconsistente.empty:
        print(f"  {col}: {inconsistente.tolist()}")
print()

for col in colunas_str:
    df_problemas[col] = df_problemas[col].str.strip().str.title()

print("Strings padronizadas com strip() e title().")
```

```python
df_problemas['Preco_Unitario'] = df_problemas['Vendas'] / df_problemas['Quantidade']
df_problemas['Faixa_Vendas'] = pd.cut(df_problemas['Vendas'], bins=3, labels=['Baixo', 'Médio', 'Alto'])
df_problemas['Meta_Atingida'] = np.where(df_problemas['Vendas'] >= 5000, 'Sim', 'Não')
df_problemas['Fat_Vendedor'] = df_problemas.groupby('Vendedor')['Vendas'].transform('sum')
df_problemas['Rank_Vendas'] = df_problemas['Vendas'].rank(ascending=False).astype(int)

print(f"Preço unitário médio: R$ {df_problemas['Preco_Unitario'].mean():.2f}")
print(f"Vendas que atingiram a meta: {df_problemas['Meta_Atingida'].value_counts().get('Sim', 0)} de {len(df_problemas)}")
print(f"Vendedor com maior faturamento total: {df_problemas.groupby('Vendedor')['Vendas'].sum().idxmax()}")
```

## Desafio 3: Visualização

Crie gráficos que mostrem:
- Comparação entre vendedores
- Tendências por região
- Distribuição dos valores

```python
import matplotlib.pyplot as plt
```

```python
vendedores = df_problemas.groupby('Vendedor')['Vendas'].sum().sort_values(ascending=True)

vendedores.plot(kind='barh', color=['#4C72B0', '#55A868', '#C44E52'])
plt.title('Comparação de Faturamento entre Vendedores')
plt.xlabel('Faturamento Total (R$)')
plt.ylabel('Vendedor')
plt.tight_layout()
plt.show()
```

```python
regioes = df_problemas.groupby('Regiao')['Vendas'].sum().sort_values(ascending=False)

regioes.plot(kind='bar', color=['#4C72B0', '#55A868', '#C44E52', '#8172B2', '#CCB974'])
plt.title('Faturamento Total por Região')
plt.xlabel('Região')
plt.ylabel('Faturamento Total (R$)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(df_problemas['Vendas'].dropna(), bins=10, color='#4C72B0', edgecolor='black')
axes[0].set_title('Distribuição de Vendas')
axes[0].set_xlabel('Valor da Venda (R$)')
axes[0].set_ylabel('Frequência')

axes[1].hist(df_problemas['Quantidade'].dropna(), bins=10, color='#55A868', edgecolor='black')
axes[1].set_title('Distribuição de Quantidade')
axes[1].set_xlabel('Quantidade')
axes[1].set_ylabel('Frequência')

plt.tight_layout()
plt.show()
```
